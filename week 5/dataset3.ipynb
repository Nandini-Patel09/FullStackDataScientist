```python
import numpy as np
import pandas as pd               
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_breast_cancer

sns.set_theme(style="whitegrid")
```

```python
from sklearn.datasets import load_breast_cancer
cancer = load_breast_cancer()
df = pd.DataFrame(cancer.data, columns=cancer.feature_names)
df['target'] = cancer.target
df.head()
```

```python
print(df.shape)
print(df.isnull().sum())
print(df.describe())
```

```python
avg_mean_radius = np.mean(df['mean radius'])
print("Average Mean Radius:", avg_mean_radius)

mean_by_target = df.groupby('target')['mean radius'].mean()
print(mean_by_target)
```

```python
plt.bar(mean_by_target.index, mean_by_target.values, color=['pink','lightblue'])
plt.title("Mean Radius by Cancer Type")
plt.xlabel("Target (0=Malignant, 1=Benign)")
plt.ylabel("Mean Radius")
plt.show()
```

```python
sns.histplot(df['mean radius'], kde=True, bins=15)
plt.title("Distribution of Mean Radius")
plt.show()

sns.heatmap(df.corr().iloc[:10, :10], annot=True, cmap='coolwarm')
plt.title("Partial Correlation Heatmap")
plt.show()
```
