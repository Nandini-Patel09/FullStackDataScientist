# 1. Nominal Variables

These variables represent labels or categories with no inherent order.

Loan_ID

Gender

Married

Education

Self_Employed

Property_Area

Loan_Status (Approved)

# 2. Ordinal Variables

These variables represent categories with a meaningful order or ranking.

Dependents (0 < 1 < 2 < 3+)

Credit_History (0 < 1 → bad → good)

# 3. Discrete Variables

These are numeric variables obtained by counting; they take integer values.

ApplicantIncome

CoapplicantIncome

Loan_Amount_Term

# 4. Continuous Variables

These are numeric variables measured on a continuous scale and may include decimals.

LoanAmount
