```python
import pandas as pd
import numpy as np
from scipy import stats
import statsmodels.api as sm
from statsmodels.stats.weightstats import ztest
from statsmodels.stats.multicomp import pairwise_tukeyhsd
```

```python
import pandas as pd

df = pd.read_csv("supermarket_sales.csv")

```

```python
# Z–TEST (Mean Rating vs 7)

z_stat, p_val = ztest(df['Rating'], value=7)
print("Z-Test:")
print("Z statistic =", z_stat)
print("P-value =", p_val, "\n")
```

```python
# INDEPENDENT T–TEST 
# Compare Total Purchase by Gender

male_total = df[df['Gender'] == "Male"]['Total']
female_total = df[df['Gender'] == "Female"]['Total']

t_stat, p_val = stats.ttest_ind(male_total, female_total)
print("Independent T-Test (Total vs Gender):")
print("T statistic =", t_stat)
print("P-value =", p_val, "\n")
```

```python
# PAIRED T–TEST 
# Compare Unit Price vs COGS (matched per row)

t_stat, p_val = stats.ttest_rel(df['Unit price'], df['cogs'])
print("Paired T-Test (Unit price vs COGS):")
print("T statistic =", t_stat)
print("P-value =", p_val, "\n")
```

```python
# ANOVA 
# Compare Total across Product Lines

groups = [g["Total"].values for _, g in df.groupby("Product line")]

f_stat, p_val = stats.f_oneway(*groups)
print("ANOVA (Total across Product Lines):")
print("F statistic =", f_stat)
print("P-value =", p_val, "\n")
```

```python
# TUKEY HSD POST-HOC

tukey = pairwise_tukeyhsd(
    endog=df['Total'],
    groups=df['Product line'],
    alpha=0.05
)
print("Tukey HSD Result:")
print(tukey, "\n")
```

```python
# CHI-SQUARE TEST
# Gender vs Branch

table = pd.crosstab(df['Gender'], df['Branch'])
chi2, p, dof, expected = stats.chi2_contingency(table)

print("Chi-Square Test (Gender vs Branch):")
print("Chi2 statistic =", chi2)
print("P-value =", p)
print("Degrees of freedom =", dof)
print("Expected frequencies:\n", expected)
```
